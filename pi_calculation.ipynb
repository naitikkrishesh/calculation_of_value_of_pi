```python
import numpy as np 
import random
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.uniform(-1,1,5)
```

```python
def circle(x,y):
    circle = x**2 + y**2
    return circle

def gen_x_y(n):
    # x , y = [],[]
    # for i in range(n):
    #     x.append(np.random.uniform(-1,1))
    #     y.append(np.random.uniform(-1,1))
    x = np.random.uniform(-1,1,n)
    y = np.random.uniform(-1,1,n)
    
    return x, y, circle(x, y), np.where(circle(x, y) <=1 , 1 ,0)





print(gen_x_y(5))
```

```python
n = 50000000

df = pd.DataFrame(gen_x_y(n)).T # T for Transpose

df.columns = ['x','y','x^2 + y^2', 'is_inside']
# df
```

```python
plt.scatter(df['x'], df['y'], c = np.where(df['is_inside'] == 1, 'red' , 'green' ))
plt.xlabel('X')
plt.ylabel('Y')
# plt.tight_layout()
plt.show()
```

```python
pi = 4* (np.sum(df['is_inside'])/n)

print(pi)
```
